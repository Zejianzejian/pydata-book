```python
nsteps = 100
rng = np.random.default_rng(seed=12345)  # fresh random generator
draws = rng.integers(0, 2, size=nsteps) # 生成0到 1 的整数
steps = np.where(draws == 0, 1, -1)
walk = steps.cumsum() # 累加看走到哪里了
```

```python
walk.min()
walk.max()
```

```python
max_index = (np.abs(walk) >= 5).argmax() # 拿到最大值的坐标，如果有重的就拿取第一个
np.abs(walk) >= 5 # 绝对值大于5
max_index
walk[max_index]
```

```python
nwalks = 5000
nsteps = 1000
draws = rng.integers(0, 2, size=(nwalks, nsteps)) # 0 or 1 生成nwalks * nsteps的矩阵
steps = np.where(draws > 0, 1, -1)
steps
walks = steps.cumsum(axis=1)
walks
```

```python
walks.max()
walks.min()
```

```python
# any tests whether one or more values in an array is True, while all checks if every value is True
np.abs(walks) >= 30
hits30_x=(np.abs(walks) >= 30).any(axis=0) # 沿着列检查是否存在至少一个 True 值，即 0 轴，每一个列都是 0 轴上数.size 为 nsteps
hits30_x.argmax() # 所有walks中最快 63 步有大于 30 的
hits30 = (np.abs(walks) >= 30).any(axis=1) # 注意是绝对值
hits30
hits30.sum() # Number that hit 30 or -30
```

```python
crossing_times = (np.abs(walks[hits30]) >= 30).argmax(axis=1) #  每个 walk最早大于30的index
crossing_times.min() # 这里印证上了
```

```python
crossing_times.mean()
```
